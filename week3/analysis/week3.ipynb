**Outlier Verification**

```python
for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col)
    print("Outliers:", len(outliers))
```

### Business Insight: Outlier & Variability Management
- **Fee Skewness**: With 1,208 outliers in fees, the platform hosts a wide variety of economic tiers.
- **Scholarship Variance**: The high number of outliers in `microscholarship` (1,641) indicates that while there is a standard rate, there are significant "premium" or "specialty" incentives that may need distinct marketing strategies.

## Executive Summary & Final EDA Conclusion

### 1. Data Integrity & Governance
- **Significant Information Gaps**: With over **90% missing data in archiving** and **81% in role responsibilities**, the platform currently lacks the granular data needed for deep matching or historical trend analysis.
- **Technical Anomalies**: The discovery of **361 records** with application deadlines preceding their creation dates suggests an urgent need for backend validation logic to prevent user frustration.

### 2. Market Positioning
- **Core Strengths**: The platform has successfully carved out a niche in the **Internship and Career** segments, which together represent nearly half of all offerings.
- **Remote-First Identity**: The near-total dominance of **Virtual/Work From Home** locations positions the platform as a key player for remote engagement, though it limits growth in industries requiring a physical presence.

### 3. Economic Model
- **Incentive-Driven**: The business model leans heavily toward **free access** with standardized low-value **microscholarships (~$120)**. This suggests a high-volume, low-barrier-to-entry strategy designed to maximize student engagement.
- **High-End Outliers**: A small subset of high-fee and high-scholarship opportunities exists, which may require a segmented marketing approach to distinguish them from the standard free offerings.

### 4. Growth Trajectory
- The platform experienced an exponential **85x growth in listings** between 2022 and 2023. Maintaining this momentum will require improving data quality standards to ensure that the increased quantity of opportunities does not come at the cost of relevance and accuracy.
